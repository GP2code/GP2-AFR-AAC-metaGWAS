# GWAS 
1. Generate scripts and submit
2. Concat results per chr into single file
3. Quick check any variants P<5E-8

## GP2 AAC

```python
%%bash

cd $WORK_DIR
mkdir -p ${WORK_DIR}/results/GP2_R11/AAC
mkdir -p ${WORK_DIR}/swarms/logs/plink_AAC

for chrom in {1..22}; do
  echo "plink2 --pfile ${DATA_DIR}/GP2_R11/AAC/chr${chrom}_AAC_release11_vwb \
    --glm hide-covar no-x-sex firth-fallback cols=+a1freq,+a1freqcc,+a1count,+totallele,+a1countcc,+totallelecc,+err \
    --pheno-name PHENO --covar-variance-standardize \
    --pheno ${DATA_DIR}/GP2_R11/GP2_R11_AFR_AAC_covariates_noRelateds_CC_PCs.tsv \
    --covar ${DATA_DIR}/GP2_R11/GP2_R11_AFR_AAC_covariates_noRelateds_CC_PCs.tsv \
    --covar-name SEX,PC1,PC2,PC3,PC4,PC5,PC6,PC7,PC8,PC9,PC10,PC11,PC12,PC13,PC14,PC15,PC16,PC17,PC18,PC19,PC20 \
    --threads 4 \
    --out ${WORK_DIR}/results/GP2_R11/AAC/chr${chrom}_AAC_release11_vwb" >> ${WORK_DIR}/swarms/GP2_R11_AAC.swarm
done
```

```python
%%bash
cd ${WORK_DIR}/swarms/

swarm -f ${WORK_DIR}/swarms/GP2_R11_AAC.swarm \
-t 4 \
--module plink \
-g 400 --time 10:00:00 \
--logdir ${WORK_DIR}/swarms/logs/plink_AAC
```

```python
%%bash

## AAC
cd ${WORK_DIR}/results/GP2_R11/AAC

head -n 1 chr1_AAC_release11_vwb.PHENO.glm.logistic.hybrid > AAC_release11_vwb.PHENO.hybrid.ALLCHR.filtered.tsv

# Append only where ERRCODE == . 
for i in {1..22}; do
echo "processing chr${i}"
awk 'NR>1 && $NF=="."' chr${i}_AAC_release11_vwb.PHENO.glm.logistic.hybrid >> AAC_release11_vwb.PHENO.hybrid.ALLCHR.filtered.tsv
done
```

```python
! wc -l {WORK_DIR}/results/GP2_R11/AAC/AAC_release11_vwb.PHENO.hybrid.ALLCHR.filtered.tsv
```

## GP2 AFR

```python
%%bash

WORK_DIR=/data/CARD_AA/projects/2025_2026_AFR_AAC_GWAS
DATA_DIR=/data/CARD_AA/projects/2025_2026_AFR_AAC_GWAS/data

cd $WORK_DIR
mkdir -p ${WORK_DIR}/results/GP2_R11/AFR
mkdir -p ${WORK_DIR}/swarms/logs/plink_AFR

for chrom in {1..22}; do
  echo "plink2 --pfile ${DATA_DIR}/GP2_R11/AFR/chr${chrom}_AFR_release11_vwb \
    --glm hide-covar no-x-sex firth-fallback cols=+a1freq,+a1freqcc,+a1count,+totallele,+a1countcc,+totallelecc,+err \
    --pheno-name PHENO --covar-variance-standardize \
    --pheno ${DATA_DIR}/GP2_R11/GP2_R11_AFR_AAC_covariates_noRelateds_CC_PCs.tsv \
    --covar ${DATA_DIR}/GP2_R11/GP2_R11_AFR_AAC_covariates_noRelateds_CC_PCs.tsv \
    --covar-name SEX,PC1,PC2,PC3,PC4,PC5,PC6,PC7,PC8,PC9,PC10,PC11,PC12,PC13,PC14,PC15,PC16,PC17,PC18,PC19,PC20 \
    --threads 4 \
    --out ${WORK_DIR}/results/GP2_R11/AFR/chr${chrom}_AFR_release11_vwb" >> ${WORK_DIR}/swarms/GP2_R11_AFR.swarm
done
```

```python
%%bash
cd ${WORK_DIR}/swarms/

swarm -f ${WORK_DIR}/swarms/GP2_R11_AFR.swarm \
-t 4 \
--module plink \
-g 400 --time 10:00:00 \
--logdir ${WORK_DIR}/swarms/logs/plink_AFR
```

```python
%%bash

## AFR
cd ${WORK_DIR}/results/GP2_R11/AFR

head -n 1 chr1_AFR_release11_vwb.PHENO.glm.logistic.hybrid > AFR_release11_vwb.PHENO.hybrid.ALLCHR.filtered.tsv

# Append only where ERRCODE == . 
for i in {1..22}; do
echo "processing chr${i}"
awk 'NR>1 && $NF=="."' chr${i}_AFR_release11_vwb.PHENO.glm.logistic.hybrid >> AFR_release11_vwb.PHENO.hybrid.ALLCHR.filtered.tsv
done
```

```python
! wc -l {WORK_DIR}/results/GP2_R11/AFR/AFR_release11_vwb.PHENO.hybrid.ALLCHR.filtered.tsv
```

## Investigate SNPs P<5E-8 

```python
%%bash

## AAC
awk 'NR==1 || $24 < 5e-8' ${WORK_DIR}/results/GP2_R11/AAC/AAC_release11_vwb.PHENO.hybrid.ALLCHR.filtered.tsv | wc -l
awk 'NR==1 || $24 < 5e-8' ${WORK_DIR}/results/GP2_R11/AAC/AAC_release11_vwb.PHENO.hybrid.ALLCHR.filtered.tsv > ${WORK_DIR}/results/GP2_R11/AAC/AAC_release11_vwb.P5E-8.tsv
cat ${WORK_DIR}/results/GP2_R11/AAC/AAC_release11_vwb.P5E-8.tsv | cut -f 1-5,24
```
